```python
import pandas as pd
```

```python
df = pd.read_excel('Dataset for Data Analytics (1).xlsx')
```

```python
df.describe()
```

```python
product_sales = df.groupby('Product')['TotalPrice'].sum()
product_sales.sort_values(ascending=False)
```

```python
import matplotlib.pyplot as plt

product_sales.plot(kind='bar')

plt.title('Product-wise Sales')
plt.xlabel('Products')
plt.ylabel('Sales')

plt.show()
```

```python
df['OrderStatus'].value_counts()
```

```python
df['OrderStatus'].value_counts().plot(kind='pie', autopct='%1.1f%%')

plt.title('Order Status')
plt.show()
```

```python
df['PaymentMethod'].value_counts().plot(kind='bar',color="yellow")

plt.title('Payment Methods')
plt.show()
```
