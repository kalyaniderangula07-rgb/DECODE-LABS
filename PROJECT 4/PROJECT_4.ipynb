```python
import pandas as pd
```

```python
df = pd.read_excel('Dataset.xlsx')
```

#1. Product-wise Sales (Bar Chart)

```python
import matplotlib.pyplot as plt

product_sales = df.groupby('Product')['TotalPrice'].sum()

plt.figure(figsize=(10,6))

bars = plt.bar(
    product_sales.index,
    product_sales.values,
    color='red',
    edgecolor='yellow',
    linewidth=2
)

plt.title("Product-wise Sales", fontsize=18, fontweight='bold')
plt.xlabel("Products", fontsize=13)
plt.ylabel("Total Sales", fontsize=13)

plt.grid(axis='y', linestyle='--', alpha=0.5)

for bar in bars:
    plt.text(bar.get_x()+bar.get_width()/2,
             bar.get_height()+1500,
             f'{int(bar.get_height())}',
             ha='center',
             fontsize=10)

plt.xticks(rotation=30)

plt.show()
```

Insight: Shows CHAIRS AND PRINTERS generate the highest revenue.

#2. Order Status Distribution (Pie Chart)

```python
plt.figure(figsize=(7,7))

colors = ['yellow','red','purple','violet','skyblue']

df['OrderStatus'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    startangle=90,
    colors=colors,
    shadow=True,
    explode=[0.05,0.05,0.05,0.05,0.05],
    wedgeprops={'edgecolor':'black'}
)

plt.title("Order Status Distribution",fontsize=18,fontweight='bold')
plt.ylabel("")

plt.show()
```

Insight: Displays the percentage of delivered, shipped, cancelled, pending, and returned orders.
Mostly were CANCELLED and RETURNED

#3. Payment Methods

```python
payment = df['PaymentMethod'].value_counts()

plt.figure(figsize=(8,6))

bars = plt.bar(
    payment.index,
    payment.values,
    color=['gold','tomato','limegreen','deepskyblue','violet'],
    edgecolor='black',
    linewidth=2
)

plt.title("Payment Methods",fontsize=18,fontweight='bold')
plt.xlabel("Payment Method")
plt.ylabel("Number of Orders")

plt.grid(axis='y',linestyle='--',alpha=0.4)

for bar in bars:
    plt.text(bar.get_x()+bar.get_width()/2,
             bar.get_height()+2,
             bar.get_height(),
             ha='center')

plt.show()
```

Insight:Most preferred payment method is ONLINE PAYMENT!

#4. Monthly Sales Trend

```python
df['Date'] = pd.to_datetime(df['Date'])

monthly = df.groupby(df['Date'].dt.month)['TotalPrice'].sum()

plt.figure(figsize=(10,6))

plt.plot(
    monthly.index,
    monthly.values,
    marker='o',
    linewidth=3,
    markersize=9,
    color='purple'
)

plt.fill_between(monthly.index, monthly.values,
                 color='violet', alpha=0.2)

plt.title("Monthly Sales Trend",fontsize=18,fontweight='bold')
plt.xlabel("Month")
plt.ylabel("Sales")

plt.grid(True,linestyle='--',alpha=0.5)

plt.show()
```

Insight: Shows how sales change over time.

#5. Quantity Distribution

```python
plt.figure(figsize=(8,6))

plt.hist(
    df['Quantity'],
    bins=10,
    color='orange',
    edgecolor='black'
)

plt.title("Quantity Distribution",fontsize=18,fontweight='bold')
plt.xlabel("Quantity")
plt.ylabel("Frequency")

plt.grid(axis='y',linestyle='--',alpha=0.5)

plt.show()
```

Insight: Shows how frequently different purchase quantities occur.

#6. Total Price Box Plot

```python
plt.figure(figsize=(6,7))

plt.boxplot(
    df['TotalPrice'],
    patch_artist=True,
    boxprops=dict(facecolor='lightgreen',
                  color='black',
                  linewidth=2),
    whiskerprops=dict(color='blue',
                      linewidth=2),
    capprops=dict(color='red',
                  linewidth=2),
    medianprops=dict(color='darkred',
                     linewidth=3),
    flierprops=dict(marker='o',
                    markerfacecolor='red',
                    markersize=8)
)

plt.title("Total Price Distribution",fontsize=18,fontweight='bold')
plt.ylabel("Total Price")

plt.grid(True,linestyle='--',alpha=0.4)

plt.show()
```

Insight: Detects high-value outliers in transaction amounts.

###Key Insights to Include
Printer and Chair generated the highest total sales.


Online payment was the most frequently used payment method.


Order statuses were fairly evenly distributed.


Monthly sales show how revenue changes over time.

Most customers purchased only a few items per order.

A small number of high-value transactions appear as outliers.
